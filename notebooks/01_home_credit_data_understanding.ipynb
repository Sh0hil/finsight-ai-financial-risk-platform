```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

df = pd.read_csv("../data/raw/home_credit/application_train.csv")

df.head()
```

```python
print("Shape of dataset:", df.shape)
df.info()
```

```python
df["TARGET"].value_counts()
```

```python
df["TARGET"].value_counts(normalize=True) * 100
```

```python
missing_values = df.isnull().mean().sort_values(ascending=False) * 100
missing_values.head(30)
```

```python
df.describe()
```

```python
df["CREDIT_INCOME_RATIO"] = df["AMT_CREDIT"] / df["AMT_INCOME_TOTAL"]
df["ANNUITY_INCOME_RATIO"] = df["AMT_ANNUITY"] / df["AMT_INCOME_TOTAL"]
df["CREDIT_TERM"] = df["AMT_ANNUITY"] / df["AMT_CREDIT"]
df["DAYS_EMPLOYED_RATIO"] = df["DAYS_EMPLOYED"] / df["DAYS_BIRTH"]

df[[
    "AMT_INCOME_TOTAL",
    "AMT_CREDIT",
    "AMT_ANNUITY",
    "CREDIT_INCOME_RATIO",
    "ANNUITY_INCOME_RATIO",
    "CREDIT_TERM",
    "DAYS_EMPLOYED_RATIO"
]].head()
```

```python
sns.countplot(data=df, x="TARGET")
plt.title("Credit Default Target Distribution")
plt.xlabel("Target: 0 = No Default, 1 = Default Risk")
plt.ylabel("Number of Applicants")
plt.show()
```

```python
sns.boxplot(data=df, x="TARGET", y="AMT_INCOME_TOTAL")
plt.title("Income Distribution by Default Risk")
plt.ylim(0, 500000)
plt.show()
```

```python
sns.boxplot(data=df, x="TARGET", y="AMT_CREDIT")
plt.title("Credit Amount Distribution by Default Risk")
plt.show()
```

```python
contract_default = df.groupby("NAME_CONTRACT_TYPE")["TARGET"].mean().sort_values(ascending=False)

contract_default.plot(kind="bar")
plt.title("Default Rate by Contract Type")
plt.ylabel("Default Rate")
plt.show()
```

```python
selected_columns = [
    "TARGET",
    "NAME_CONTRACT_TYPE",
    "CODE_GENDER",
    "FLAG_OWN_CAR",
    "FLAG_OWN_REALTY",
    "CNT_CHILDREN",
    "AMT_INCOME_TOTAL",
    "AMT_CREDIT",
    "AMT_ANNUITY",
    "AMT_GOODS_PRICE",
    "NAME_INCOME_TYPE",
    "NAME_EDUCATION_TYPE",
    "NAME_FAMILY_STATUS",
    "NAME_HOUSING_TYPE",
    "DAYS_BIRTH",
    "DAYS_EMPLOYED",
    "OCCUPATION_TYPE",
    "CNT_FAM_MEMBERS",
    "CREDIT_INCOME_RATIO",
    "ANNUITY_INCOME_RATIO",
    "CREDIT_TERM",
    "DAYS_EMPLOYED_RATIO"
]

credit_clean = df[selected_columns].copy()

credit_clean.to_csv("../data/processed/credit_risk_base.csv", index=False)

print("Saved successfully!")
print(credit_clean.shape)
```
